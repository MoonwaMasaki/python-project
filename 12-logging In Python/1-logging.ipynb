## Python Logging

Logging is crucial aspect of any application providing a way to track events, errors and operational information. Python built-in logging module offers a flexible framework for emitting log messages from Python programs. 

```python
import logging 

# Configure the basic logging settings
logging.basicConfig(level=logging.DEBUG)

# log messages with different severity levels
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

### Log Levels

Python's logging module has several log levels indicating the severity of events. The default levels are:

- DEBUG: Detailed information, typically of interest only when diagnosing problems.
- INFO: Confirmation that things are working as expected
- WARNING: An indication that something unexpected happened or indicative of some problem in the near future(eg. 'disk space low'). The software is still working as expected
- ERROR: Due to a more serious problem, the software has not been able to perform some functions
- CRITICAL: A very serious error, indicating that the problem itself may be unable to continue running

```python
## Configuring Logging Levels
# You can set the logging level to control which messages are logged
import logging

logging.basicConfig(
    filename='app.log',  # Log messages will be written to this file
    filemode='w',  # 'w' to overwrite the file, 'a' to append
    level=logging.DEBUG,
    FORMAT='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
    )  # Set to INFO level

# log messages with different severity levels
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```
